```python
from pathlib import Path
import boto3

NETID = "zw520"
REGION = "us-east-1"

bucket_name = f"dsan6000-{NETID}"

s3 = boto3.client("s3", region_name=REGION)
```

```python
if REGION == "us-east-1":
    result = s3.create_bucket(Bucket=bucket_name)
else:
    result = s3.create_bucket(
        Bucket=bucket_name,
        CreateBucketConfiguration={
            "LocationConstraint": REGION
        }
    )

result
```

```python
for file in sorted(Path("data").glob("*.parquet")):
    s3.upload_file(
        str(file),
        bucket_name,
        f"wikipedia-hourly/{file.name}"
    )
```

```python
response = s3.list_objects_v2(
    Bucket=bucket_name,
    Prefix="wikipedia-hourly/"
)

sum(
    item["Key"].endswith(".parquet")
    for item in response.get("Contents", [])
)
```
